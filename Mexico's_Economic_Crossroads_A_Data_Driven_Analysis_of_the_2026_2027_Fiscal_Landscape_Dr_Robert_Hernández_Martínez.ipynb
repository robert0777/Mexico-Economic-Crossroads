# **Mexico's Economic Crossroads**
## A Data-Driven Analysis of the 2026–2027 Fiscal Landscape

### Executive Summary

As of mid-2026, the Mexican economy is navigating a structural downshift defined by regulatory volatility and fiscal compression. On May 22, 2026, INEGI confirmed that the national economy contracted by 0.6% quarter-over-quarter in Q1 2026, marking the weakest start to a year since the 2020 pandemic shock.

This is not a minor dip. While the Ministry of Finance (SHCP) continues to hold onto an unrealistic 2.3% growth target, independent revisions tell a different story. The Central Bank (Banxico) cut its projection to 1.1% (with a lower boundary of 0.5%), and the July 7, 2026, Citi Mexico Expectations Survey shows 11 institutions forecasting growth below 1.0%.

For finance professionals and individuals alike, the core objective for the next 18 months must shift from aggressive expansion to defensive capital preservation and high liquidity.

To quantify this risk, a Composite Crisis Probability Model (CCPM) was constructed using a KLR Signals-Based Monte Carlo simulation with 100,000 iterations. The model is grounded in three established academic frameworks: the KLR signals approach, the Berg-Pattillo Probit/Logit framework, and the World Bank Panel Logit methodology.

The model yields a point estimate of 74.1%. The 80% confidence interval for the probability of a severe economic crisis in Mexico within the next 18 months is [54.91%, 85.65%]. The distribution is concentrated in the high-probability region: there is a 94.56% chance that the true probability exceeds 50%.

The hazard rate is not evenly distributed across the 18-month horizon. The data shows it peaks in February 2027, driven by the sequential collision of post-World Cup revenue collapse, year-end debt service peaks, and the inaugural USMCA annual review.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import copy
import time

# ============================================
# 1. SHARED CONFIGURATION
# ============================================
n_simulations = 100_000
batch_size = 2000
n_batches = n_simulations // batch_size
np.random.seed(42)

# ============================================
# 2. KLR INDICATORS
# ============================================
indicators = {
    'GDP_Growth': {
        'current': 1.2,
        'score_map': [(0.0, 3.0), (1.0, 2.5), (2.0, 1.5), (3.0, 0.5)],
        'weight': 0.20,
        'source': 'IMF WEO July 2026 Update / INEGI Q1 Trajectory',
        'description': 'Annual GDP growth (%). Downgraded due to trade/energy bottlenecks.'
    },
    'Core_Inflation': {
        'current': 3.9,
        'score_map': [(5.0, 3.0), (4.5, 2.5), (4.0, 2.0), (3.0, 0.5)],
        'weight': 0.10,
        'source': 'IMF July 2026 / INEGI INPC Baseline',
        'description': 'Core inflation (%). Marginally lower but structural pressures remain.'
    },
    'Sovereign_Rating': {
        'current': 2.5,
        'score_map': [(3.0, 3.0), (2.5, 2.5), (2.0, 1.5), (1.0, 0.5)],
        'weight': 0.20,
        'source': "Moody's Investors Service, May 20 2026",
        'description': 'Sovereign credit rating mapped to numerical scale.'
    },
    'Banking_Sector': {
        'current': 3.0,
        'score_map': [(3.0, 3.0), (2.0, 2.0), (1.0, 1.0), (0.0, 0.0)],
        'weight': 0.15,
        'source': "Moody's Investors Service, May 21 2026",
        'description': 'Systemic banking sector credit stress.'
    },
    'Fiscal_Deficit': {
        'current': 5.4,
        'score_map': [(6.0, 3.0), (5.0, 2.5), (4.0, 2.0), (3.0, 0.5)],
        'weight': 0.15,
        'source': 'SHCP / CIEP Paquete Económico 2026',
        'description': 'Federal deficit (% GDP). Exceeds 3.5% sustainable threshold.'
    },
    'Debt_Service': {
        'current': 18.0,
        'score_map': [(25.0, 3.0), (20.0, 2.5), (15.0, 2.0), (10.0, 0.5)],
        'weight': 0.10,
        'source': 'SHCP Debt Sustainability Matrices',
        'description': 'Interest payments as % of total federal revenue.'
    },
    'USMCA_Uncertainty': {
        'current': 2.0,
        'score_map': [(3.0, 3.0), (2.0, 2.0), (1.0, 1.0), (0.0, 0.0)],
        'weight': 0.05,
        'source': 'USTR July 1 2026 / White & Case LLP July 2 2026',
        'description': 'Trade agreement uncertainty. 0=Renewed, 1=Review, 2=Annual reviews, 3=Withdrawal.'
    },
    'Institutional_Quality': {
        'current': 2.0,
        'score_map': [(3.0, 3.0), (2.0, 2.0), (1.0, 1.0), (0.0, 0.0)],
        'weight': 0.05,
        'source': 'Transparency International Feb 2026 / OECD March 2026',
        'description': 'Corruption + bureaucratic risk composite.'
    }
}

# ============================================
# 3. SHARED FUNCTIONS
# ============================================
def calculate_score(indicator):
    """Map current value to crisis score using piecewise linear interpolation."""
    val = indicator['current']
    score_map = sorted(indicator['score_map'], key=lambda x: x[0])
    if val >= score_map[-1][0]:
        return score_map[-1][1]
    if val <= score_map[0][0]:
        return score_map[0][1]
    for i in range(len(score_map)-1):
        x0, y0 = score_map[i]
        x1, y1 = score_map[i+1]
        if x0 <= val <= x1:
            return y0 + (y1 - y0) * (val - x0) / (x1 - x0)
    return 0.0

def compute_composite_index(indicators_dict, weight_noise=0.0, score_noise=0.0):
    """Compute Composite Crisis Index (CCI) normalized to [0,1]."""
    total_weight = sum(ind['weight'] for ind in indicators_dict.values())
    cci = 0.0
    scores = {}
    for name, ind in indicators_dict.items():
        w = ind['weight'] * (1 + weight_noise)
        s = calculate_score(ind) + score_noise
        s = np.clip(s, 0, 3)
        scores[name] = s
        cci += w * s
    cci = cci / (total_weight * 3.0)
    return np.clip(cci, 0, 1), scores

def crisis_probability_decimal(cci, alpha=5.0, beta=0.60):
    """Logistic transformation returning decimal probability (0.0 to 1.0)."""
    return 1.0 / (1.0 + np.exp(-alpha * (cci - beta)))

def crisis_probability_percent(cci, alpha=5.0, beta=0.60):
    """Logistic transformation returning percentage (0.0 to 100.0)."""
    return (1.0 / (1.0 + np.exp(-alpha * (cci - beta)))) * 100.0

def calculate_real_estate_trajectory(cci, months=13):
    """
    Calculate real estate index trajectory based on CCI.
    Uses the quadratic formula from the technical appendix.
    """
    P0 = 100.0
    c_m = 0.0125   # Credit Compression Variable
    i_s = 0.0065   # Infrastructure & Safety Risk Multiplier
    phi = 0.00122  # Second-Order Market Stabilization Coefficient

    # Scale risk factors by CCI
    effective_cm = c_m * (0.8 + 0.4 * cci)
    effective_is = i_s * (0.7 + 0.6 * cci)

    t = np.arange(months)
    trajectory = P0 * (1.0 - (effective_cm + effective_is) * t + phi * (t**2))
    return trajectory

# ============================================
# 4. POINT ESTIMATE (Deterministic)
# ============================================
print("=" * 70)
print("POINT ESTIMATE (Deterministic)")
print("=" * 70)

point_cci, point_scores = compute_composite_index(indicators)
point_prob_decimal = crisis_probability_decimal(point_cci)      # 0.741
point_prob_percent = crisis_probability_percent(point_cci)      # 74.1

# Calculate the weighted sum for transparency
total_weight = sum(ind['weight'] for ind in indicators.values())
weighted_sum = sum(ind['weight'] * calculate_score(ind) for ind in indicators.values())

print(f"Weighted Sum (Σ Wi × Si): {weighted_sum:.3f}")
print(f"Total Weight (Σ Wi): {total_weight:.2f}")
print(f"Point CCI: {point_cci:.4f}")
print(f"Point Probability: {point_prob_percent:.2f}%")

# Display individual scores
print("\n" + "-" * 70)
print("Indicator Scores (Point Estimate):")
print("-" * 70)
for name, score in point_scores.items():
    w = indicators[name]['weight']
    contrib = w * score
    print(f"{name:<25} Score: {score:.2f}  Weight: {w:.2f}  Contrib: {contrib:.3f}")

# ============================================
# 5. UNIFIED MONTE CARLO SIMULATION
# ============================================
print("\n" + "=" * 70)
print("UNIFIED MONTE CARLO SIMULATION (100,000 runs)")
print("=" * 70)

# Storage arrays
probabilities_decimal = []  # Store as decimals (0.0 to 1.0)
probabilities_percent = []  # Store as percentages (0.0 to 100.0)
cci_values = []
trajectories = []

months = ['May 26', 'Jun 26', 'Jul 26', 'Aug 26', 'Sep 26', 'Oct 26',
          'Nov 26', 'Dec 26', 'Jan 27', 'Feb 27', 'Mar 27', 'Apr 27', 'May 27']
t = np.arange(len(months))

for batch in tqdm(range(n_batches), desc="Running Unified Simulation", unit="batch"):
    for _ in range(batch_size):
        # Random perturbations
        w_noise = np.random.normal(0, 0.15)

        perturbed = copy.deepcopy(indicators)
        for name in perturbed:
            s_noise = np.random.normal(0, 0.25)
            perturbed[name]['current'] = max(0, perturbed[name]['current'] + s_noise)

        # Calculate CCI and Probability
        cci, _ = compute_composite_index(perturbed, weight_noise=w_noise, score_noise=0)
        prob_dec = crisis_probability_decimal(cci)
        prob_pct = prob_dec * 100.0

        # Calculate Real Estate Trajectory
        trajectory = calculate_real_estate_trajectory(cci, months=len(months))

        cci_values.append(cci)
        probabilities_decimal.append(prob_dec)
        probabilities_percent.append(prob_pct)
        trajectories.append(trajectory)

# Convert to numpy arrays
probabilities_decimal = np.array(probabilities_decimal)
probabilities_percent = np.array(probabilities_percent)
cci_values = np.array(cci_values)
trajectories = np.array(trajectories)

# ============================================
# 6. CRISIS PROBABILITY STATISTICS
# ============================================
print("\n" + "=" * 70)
print("CRISIS PROBABILITY STATISTICS")
print("=" * 70)

mean_mc_prob_decimal = np.mean(probabilities_decimal)
mean_mc_prob_percent = np.mean(probabilities_percent)
median_mc_prob_percent = np.median(probabilities_percent)
std_mc_prob_percent = np.std(probabilities_percent)
p10_mc_prob_percent = np.percentile(probabilities_percent, 10)
p90_mc_prob_percent = np.percentile(probabilities_percent, 90)

print(f"Mean Crisis Probability:      {mean_mc_prob_percent:.2f}%")
print(f"Median Crisis Probability:    {median_mc_prob_percent:.2f}%")
print(f"Standard Deviation:           {std_mc_prob_percent:.2f}%")
print(f"\nConfidence Intervals:")
print(f"  95% CI: [{np.percentile(probabilities_percent, 2.5):.2f}%, {np.percentile(probabilities_percent, 97.5):.2f}%]")
print(f"  90% CI: [{np.percentile(probabilities_percent, 5.0):.2f}%, {np.percentile(probabilities_percent, 95.0):.2f}%]")
print(f"  80% CI: [{p10_mc_prob_percent:.2f}%, {p90_mc_prob_percent:.2f}%]")
print(f"\nTail Probabilities:")
print(f"  P(Crisis) > 50%: {np.mean(probabilities_percent > 50.0):.2%}")
print(f"  P(Crisis) > 70%: {np.mean(probabilities_percent > 70.0):.2%}")
print(f"  P(Crisis) > 80%: {np.mean(probabilities_percent > 80.0):.2%}")

# ============================================
# 7. REAL ESTATE TRAJECTORY STATISTICS
# ============================================
print("\n" + "=" * 70)
print("REAL ESTATE INDEX STATISTICS")
print("=" * 70)

mean_trajectory = np.mean(trajectories, axis=0)
median_trajectory = np.median(trajectories, axis=0)
p10_trajectory = np.percentile(trajectories, 10, axis=0)
p90_trajectory = np.percentile(trajectories, 90, axis=0)

# National average (deterministic baseline)
national_average = 100.0 * (1.0 - (0.008 + 0.004) * t + 0.00067 * (t**2))

peak_idx = np.argmin(mean_trajectory)
print(f"Peak contraction (mean): {100 - np.min(mean_trajectory):.1f}% at {months[peak_idx]}")
print(f"Peak contraction (median): {100 - np.min(median_trajectory):.1f}%")
print(f"80% CI contraction: [{100 - np.min(p90_trajectory):.1f}%, {100 - np.min(p10_trajectory):.1f}%]")

# ============================================
# 8. VISUALIZATION: GRAPH 1 - Probability Distribution
# ============================================
print("\n" + "=" * 70)
print("GENERATING GRAPH 1: Probability Distribution")
print("=" * 70)

plt.figure(figsize=(12, 7), dpi=120)
sns.set_theme(style="whitegrid")

plt.hist(probabilities_percent, bins=60, color='#c0392b', alpha=0.75, edgecolor='white', linewidth=0.5)
plt.axvline(point_prob_percent, color='#2c3e50', linestyle='--', linewidth=2.5, label=f'Point Estimate: {point_prob_percent:.1f}%')
plt.axvline(p10_mc_prob_percent, color='#e67e22', linestyle=':', linewidth=2,
            label=f'80% CI: {p10_mc_prob_percent:.1f}% - {p90_mc_prob_percent:.1f}%')
plt.axvline(p90_mc_prob_percent, color='#e67e22', linestyle=':', linewidth=2)
plt.axvline(mean_mc_prob_percent, color='#8e44ad', linestyle='-.', linewidth=2,
            label=f'Mean: {mean_mc_prob_percent:.1f}%')
plt.xlabel('Crisis Probability (%)', fontsize=12, weight='semibold')
plt.ylabel('Frequency', fontsize=12, weight='semibold')
plt.title('Mexico 2026-2027: Distribution of Crisis Probabilities\n'
          'KLR + Berg-Pattillo + Monte Carlo (100,000 runs)',
          fontsize=14, weight='bold')
plt.legend(fontsize=10, loc='upper left')
plt.grid(True, alpha=0.3, linestyle=':')
plt.xlim(0, 100)

# Add annotation with key stats
stats_text = (
    f"Point Estimate: {point_prob_percent:.1f}%\n"
    f"Mean: {mean_mc_prob_percent:.1f}%\n"
    f"Median: {median_mc_prob_percent:.1f}%\n"
    f"Std Dev: {std_mc_prob_percent:.1f}%\n"
    f"80% CI: [{p10_mc_prob_percent:.1f}%, {p90_mc_prob_percent:.1f}%]\n"
    f"P(Crisis > 50%): {np.mean(probabilities_percent > 50.0):.1%}"
)
plt.text(0.97, 0.95, stats_text, transform=plt.gca().transAxes,
         fontsize=9, va='top', ha='right',
         bbox=dict(facecolor='white', alpha=0.9, edgecolor='#7f8c8d', boxstyle='round,pad=0.5'))

plt.tight_layout()
plt.show()


# ============================================
# 9. VISUALIZATION: GRAPH 2 - Real Estate Trajectory (WITH KLR INFO BOX)
# ============================================
print("\n" + "=" * 70)
print("GENERATING GRAPH 2: Real Estate Index Trajectories")
print("=" * 70)

plt.figure(figsize=(12, 7), dpi=120)
sns.set_theme(style="whitegrid")

# Plot mean trajectory with confidence band
plt.plot(months, mean_trajectory, color='#d9534f', linestyle='-', linewidth=2.5,
         label='Popular Working-Class Urban Sectors (Mean)')
plt.fill_between(months, p10_trajectory, p90_trajectory, color='#d9534f', alpha=0.25,
                  label='80% Confidence Band')

# Plot national average
plt.plot(months, national_average, color='#7f8c8d', linestyle='--', marker='o',
         alpha=0.8, linewidth=2.0, label='National Average Market Index')

# Highlight peak stress window
plt.axvspan('Nov 26', 'Jan 27', color='#f2dede', alpha=0.45,
            label='Peak Credit & Liquidity Stress Window')

# ============================================
# CORRECTED: Dynamic peak detection
# ============================================
# Find the actual peak contraction index (minimum value in mean_trajectory)
peak_idx = np.argmin(mean_trajectory)
peak_month = months[peak_idx]
peak_contraction_pct = 100.0 - mean_trajectory[peak_idx]

plt.annotate(
    f'Projected Peak Contraction: {peak_contraction_pct:.1f}%\n({peak_month})',
    xy=(peak_idx, mean_trajectory[peak_idx]),
    xytext=(peak_idx, 84),
    ha='center',
    fontsize=10,
    weight='bold',
    color='#c0392b',
    arrowprops=dict(arrowstyle='->', color='#c0392b', lw=1.5)
)

plt.xlabel('Macro Timeline Horizon (2026 - 2027)', fontsize=12, weight='semibold')
plt.ylabel('Index Base Value (100 = May 2026 Baseline)', fontsize=12, weight='semibold')
plt.title('Mexico 2026-2027: Real Estate Index Trajectory Simulation\n'
          'Popular Working-Class Urban Sectors vs. National Average',
          fontsize=14, weight='bold')
plt.legend(fontsize=10, loc='lower left')
plt.grid(True, alpha=0.3, linestyle=':')
plt.ylim(82, 105)


# ============================================
# ADD KLR RISK ENGINE INFO BOX (USING PERCENTAGES)
# ============================================
model_metrics_text = (
    f"[ QUANTITATIVE KLR RISK ENGINE ]\n"
    f"Baseline CCI Severity Index: {point_cci:.2f}\n"
    f"Deterministic Crisis Prob: {point_prob_percent:.1f}%\n"
    f"Monte Carlo Runs: {n_simulations:,}\n"
    f"Simulated Mean Crisis Prob: {mean_mc_prob_percent:.1f}%"
)

plt.text(0.97, 0.95, model_metrics_text, transform=plt.gca().transAxes,
         fontsize=9.5, color='#2c3e50', weight='semibold', va='top', ha='right',
         bbox=dict(facecolor='#f8f9fa', alpha=0.95, edgecolor='#7f8c8d', boxstyle='round,pad=0.6'))

# Add model parameters annotation (below the KLR box)
param_text = (
    f"Simulation Variables:\n"
    f"P0 = 100.0 | c_m = 0.0125 | i_s = 0.0065 | φ = 0.00122"
)
plt.text(0.97, 0.55, param_text, transform=plt.gca().transAxes,
         fontsize=8, color='#95a5a6', va='top', ha='right',
         bbox=dict(facecolor='white', alpha=0.85, edgecolor='none', boxstyle='round,pad=0.4'))

plt.tight_layout()
plt.show()

# ============================================
# 10. VISUALIZATION: GRAPH 3 - CCI-to-Probability Mapping
# ============================================
print("\n" + "=" * 70)
print("GENERATING GRAPH 3: CCI-to-Probability Logistic Mapping")
print("=" * 70)

plt.figure(figsize=(12, 7), dpi=120)
sns.set_theme(style="whitegrid")

# Plot logistic curve (using decimal for y-axis, but showing labels as percentages)
cci_curve = np.linspace(0, 1, 200)
prob_curve_decimal = crisis_probability_decimal(cci_curve)
prob_curve_percent = prob_curve_decimal * 100.0

plt.plot(cci_curve, prob_curve_percent, color='#3498db', linewidth=2.5, label='Logistic Mapping')

# Scatter Monte Carlo results (using percentages)
sample_size = min(1000, len(cci_values))
sample_indices = np.random.choice(len(cci_values), sample_size, replace=False)
plt.scatter(cci_values[sample_indices], probabilities_percent[sample_indices],
            alpha=0.15, c='#2980b9', s=10, label='Monte Carlo Samples')

# Mark point estimate
plt.scatter([point_cci], [point_prob_percent], c='#c0392b', s=200, zorder=5,
            edgecolor='black', linewidth=1.5,
            label=f'Point Estimate (CCI={point_cci:.3f}, P={point_prob_percent:.1f}%)')

# Highlight stress zone (in percentage terms)
plt.fill_betweenx([0, 100], 0.55, 0.65, alpha=0.15, color='orange',
                  label='Moderate-High Stress Zone')

# Add threshold lines
plt.axvline(x=0.60, color='#2c3e50', linestyle='--', alpha=0.5, linewidth=1.0)
plt.axhline(y=50.0, color='#2c3e50', linestyle='--', alpha=0.5, linewidth=1.0)

plt.xlabel('Composite Crisis Index (CCI)', fontsize=12, weight='semibold')
plt.ylabel('Crisis Probability (%)', fontsize=12, weight='semibold')
plt.title('Mexico 2026-2027: CCI-to-Probability Logistic Mapping\n'
          'Berg-Pattillo Sigmoid Function (α=5.0, β=0.60)',
          fontsize=14, weight='bold')
plt.xlim(0, 1)
plt.ylim(0, 100)
plt.grid(True, alpha=0.3, linestyle=':')
plt.legend(fontsize=10, loc='lower right')

# Add formula annotation
formula_text = (
    r'$P(Crisis) = \frac{1}{1 + e^{-5.0(CCI - 0.60)}} \times 100$' + '\n' + '\n' +
    f'Point CCI: {point_cci:.3f}\n' +
    f'Point Probability: {point_prob_percent:.1f}%\n' +
    f'Mean CCI (MC): {np.mean(cci_values):.3f}\n' +
    f'Mean Probability (MC): {mean_mc_prob_percent:.1f}%'
)
plt.text(0.03, 0.95, formula_text, transform=plt.gca().transAxes,
         fontsize=9.5, va='top', ha='left',
         bbox=dict(facecolor='white', alpha=0.9, edgecolor='#7f8c8d', boxstyle='round,pad=0.5'))

plt.tight_layout()
plt.show()

# ============================================
# 11. SUMMARY OUTPUT
# ============================================
print("\n" + "=" * 70)
print("SUMMARY OF RESULTS")
print("=" * 70)
print(f"\nCCI (Point Estimate):          {point_cci:.4f}")
print(f"Crisis Probability (Point):    {point_prob_percent:.2f}%")
print(f"Crisis Probability (Mean MC):  {mean_mc_prob_percent:.2f}%")
print(f"80% Confidence Interval:       [{p10_mc_prob_percent:.2f}%, {p90_mc_prob_percent:.2f}%]")
print(f"\nReal Estate Peak Contraction (Mean): {100 - np.min(mean_trajectory):.1f}%")
print(f"Real Estate Peak Month:         {months[np.argmin(mean_trajectory)]}")
print(f"\nNational Average at May 2027:  {national_average[-1]:.1f}")
print("=" * 70)
```
